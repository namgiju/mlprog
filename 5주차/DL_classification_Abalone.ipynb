# DL_classification_Abalone

전복 측정값으로 Rings 값을 분류합니다.


## 1. 라이브러리와 재현성
고정 시드를 사용하지만 실행 장치와 라이브러리 버전에 따라 결과가 달라질 수 있습니다.


```python
import os, sys, json, random
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder, LabelEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import (accuracy_score, f1_score, classification_report,
    ConfusionMatrixDisplay, roc_auc_score, RocCurveDisplay,
    mean_absolute_error, mean_squared_error, r2_score)

SEED = 42
tf.keras.backend.clear_session()
tf.keras.utils.set_random_seed(SEED)
np.random.seed(SEED)
EPOCHS = 50
BATCH_SIZE = 32
print('Python:', sys.version.split()[0], 'TensorFlow:', tf.__version__)
print('GPU:', tf.config.list_physical_devices('GPU'))
```

## 2. 데이터 불러오기


```python
USE_UPLOAD = False  # 직접 CSV를 업로드하려면 True로 변경 후 이 셀부터 다시 실행
CSV_PATH = 'abalone.csv'
DATA_URL = 'https://raw.githubusercontent.com/MyungKyuYi/AI-class/fe508cd0fdc46572dd300c0cb7b6f4346486b4b6/abalone.csv'
if USE_UPLOAD:
    from google.colab import files
    import io
    uploaded = files.upload()
    candidates = [n for n in uploaded if n.lower().endswith('.csv')]
    if len(candidates) != 1:
        raise ValueError('CSV 파일을 정확히 하나 선택하세요.')
    df = pd.read_csv(io.BytesIO(uploaded[candidates[0]]))
else:
    # ZIP에 동봉된 CSV 또는 Colab 왼쪽 파일 탭에 올린 CSV를 우선 사용
    df = pd.read_csv(CSV_PATH if Path(CSV_PATH).exists() else DATA_URL)
df.columns = df.columns.str.strip()
print('Shape:', df.shape)
print(df.head().to_string(index=False))
print('Missing values:', df.isna().sum().to_dict())
```

## 3. 입력 X와 정답 y


```python
required = ['Sex','Length','Diameter','Height','Whole_weight','Shucked_weight','Viscera_weight','Shell_weight','Rings']
assert set(required).issubset(df.columns), f'필요 컬럼: {required}'
df = df.dropna(subset=['Rings']).reset_index(drop=True)
X = df[required[:-1]].copy()  # id는 식별자이므로 제외
assert X['Sex'].dropna().isin(['I','M','F']).all()
# 교수님 분류 예제의 파생특성과 로그 변환을 반영합니다.
X['Whole_weight_ratio'] = X['Shucked_weight'] / X['Whole_weight'].replace(0, np.nan)
for col in ['Whole_weight','Shucked_weight','Viscera_weight','Shell_weight']:
    assert (X[col].dropna() >= 0).all()
    X[col] = np.log(1e-5 + X[col])
print(df['Rings'].value_counts().sort_index())
df['Rings'].hist(bins=29); plt.xlabel('Rings'); plt.ylabel('Count'); plt.show()

# 정답 이름 사전은 전체 데이터에 존재하는 Rings 값으로 고정합니다.
# 이는 특성의 통계 학습이 아니며, 훈련에 없는 클래스는 별도 경고합니다.
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(df['Rings']).astype('int32')
class_names = label_encoder.classes_
STRATIFY = False
```

## 4. 데이터 분할 및 전처리
전체의 약 64%/16%/20%를 훈련/검증/테스트로 사용합니다. 테스트는 최종 평가에만 사용합니다.


```python
# 먼저 분할한 뒤, 훈련 데이터에서만 대치값/표준화 기준을 학습합니다.
idx = np.arange(len(y))
trainval_idx, test_idx = train_test_split(idx, test_size=0.2,
    random_state=SEED, stratify=y if STRATIFY else None)
train_idx, val_idx = train_test_split(trainval_idx, test_size=0.2,
    random_state=SEED, stratify=y[trainval_idx] if STRATIFY else None)
y_train, y_val, y_test = y[train_idx], y[val_idx], y[test_idx]
print('Train / validation / test:', len(train_idx), len(val_idx), len(test_idx))
assert not set(train_idx) & set(test_idx)
assert not set(val_idx) & set(test_idx)
```

```python
num_cols = X.select_dtypes(include=np.number).columns.tolist()
cat_cols = X.columns.difference(num_cols).tolist()
preprocessor = ColumnTransformer([
    ('numeric', Pipeline([('impute', SimpleImputer(strategy='median')),
                          ('scale', StandardScaler())]), num_cols),
    ('category', Pipeline([('impute', SimpleImputer(strategy='most_frequent')),
        ('encode', OneHotEncoder(handle_unknown='ignore', sparse_output=False))]), cat_cols)
])
X_train = preprocessor.fit_transform(X.iloc[train_idx]).astype('float32')
X_val = preprocessor.transform(X.iloc[val_idx]).astype('float32')
X_test = preprocessor.transform(X.iloc[test_idx]).astype('float32')
assert np.isfinite(X_train).all() and np.isfinite(X_test).all()
print('Encoded input shape:', X_train.shape)
```

## 5. 분류 모델
이진분류는 Sigmoid + Binary Crossentropy, 다중분류는 Softmax + Sparse Categorical Crossentropy를 사용합니다. Sparse 손실은 원-핫 대신 0부터 시작하는 정수 클래스 인덱스를 받습니다.


```python
y_train_fit, y_val_fit = y_train, y_val
METRIC = 'accuracy'
unseen = sorted(set(y_test) - set(y_train))
print('훈련에 없는 테스트 클래스:', [class_names[i] for i in unseen])
print('희소 클래스는 성능 추정이 불안정합니다. 보고서의 support도 확인하세요.')
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=X_train.shape[1:]),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(32, activation='relu'),
    tf.keras.layers.Dense(len(class_names), activation='softmax')
])
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
              loss='sparse_categorical_crossentropy', metrics=[METRIC])
model.summary()
```

## 6. 학습
`fit()`에서 순전파, 손실 계산, 역전파, 가중치 갱신을 반복합니다. 검증 데이터로 조기 종료 시점을 선택합니다.


```python
# 검증 손실이 개선되지 않으면 중단하고 최적 가중치를 복원합니다.
callbacks = [tf.keras.callbacks.EarlyStopping(
    monitor='val_loss', patience=8, restore_best_weights=True)]
history = model.fit(X_train, y_train_fit,
    validation_data=(X_val, y_val_fit), epochs=EPOCHS,
    batch_size=BATCH_SIZE, callbacks=callbacks, verbose=2)
```

## 7. 학습 곡선
훈련 손실은 감소하지만 검증 손실이 증가하면 과적합을 의심합니다.


```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
for key, ax in zip(['loss', METRIC], axes):
    ax.plot(history.history[key], label='Train')
    ax.plot(history.history['val_' + key], label='Validation')
    ax.set(xlabel='Epoch (zero-based)', ylabel=key, title=key)
    ax.legend(); ax.grid(alpha=0.3)
plt.tight_layout(); plt.show()
```

## 8. 독립 테스트 평가


```python
probabilities = model.predict(X_test, verbose=0)
y_pred = probabilities.argmax(axis=1)
metrics = {'accuracy': float(accuracy_score(y_test, y_pred)),
           'macro_f1': float(f1_score(y_test, y_pred, labels=np.arange(len(class_names)), average='macro', zero_division=0))}
majority = np.bincount(y_train).argmax()
metrics['majority_baseline_accuracy'] = float(np.mean(y_test == majority))
print(metrics)
print(classification_report(y_test, y_pred, labels=np.arange(len(class_names)),
      target_names=[str(v) for v in class_names], zero_division=0))
fig, ax = plt.subplots(figsize=(12, 10) if len(class_names)>10 else (7, 6))
ConfusionMatrixDisplay.from_predictions(y_test, y_pred,
    labels=np.arange(len(class_names)), display_labels=class_names, ax=ax,
    cmap='Blues', include_values=len(class_names)<=10, colorbar=False)
ax.set_title('Test confusion matrix'); plt.tight_layout(); plt.show()
```
